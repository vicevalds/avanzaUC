```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
Tc = 0.45
```

```python
df_NN_k3 = pd.read_csv("data/NN_reinvent_stage_nim_pIC50_9_vs_Molport_Ro5.tsv", sep="\t")
df_NN_k3 = df_NN_k3[["compound_id", "compound_SMILES", "tanimoto"]]
df_NN_k3 = df_NN_k3.sort_values(["compound_id", "tanimoto"], ascending=[True, False])
df_NN_k3 = df_NN_k3.drop_duplicates("compound_id").reset_index(drop=True)
df_NN_k3
```

```python
df_NN_all = pd.read_csv("data/NN_reinvent_stage_nim_pIC50_9_vs_Molport_Ro5_all_overTc04.tsv", sep="\t")
df_NN_all = df_NN_all[["compound_id", "compound_SMILES", "tanimoto"]]
df_NN_all = df_NN_all.sort_values(["compound_id", "tanimoto"], ascending=[True, False])
df_NN_all = df_NN_all.drop_duplicates("compound_id").reset_index(drop=True)
df_NN_all
```

```python
df_NN_all_Tc = df_NN_all[df_NN_all["tanimoto"] >= Tc]

filename = "results/NN_reinvent_stage_nim_pIC50_9_vs_Molport_Ro5_Tc045_nok3"
df_NN_all_Tc.to_csv(f"{filename}.csv", index=False)
df_NN_all_Tc[["compound_SMILES", "compound_id"]].to_csv(f"{filename}.smi", index=False, header=False, sep="\t")
df_NN_all_Tc
```

```python
df_NN_Tc = df_NN_k3[df_NN_k3["tanimoto"] >= Tc]

filename = "results/NN_reinvent_stage_nim_pIC50_9_vs_Molport_Ro5_Tc045"
df_NN_Tc.to_csv(f"{filename}.csv", index=False)
df_NN_Tc[["compound_SMILES", "compound_id"]].to_csv(f"{filename}.smi", index=False, header=False, sep="\t")
df_NN_Tc
```

```python
plt.hist(df_NN_k3["tanimoto"], bins=100)
plt.show()
```
